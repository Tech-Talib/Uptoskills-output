# 🚦 AI-Based Traffic Congestion Level Estimation System

**Real-time traffic congestion estimation from CCTV video** using **YOLOv8 + ByteTrack**.

| Feature | Implementation |
|---|---|
| Detection | YOLOv8s (Ultralytics) — Car, Bus, Truck, Motorcycle, Bicycle only |
| Tracking | ByteTrack (via `supervision`) — persistent unique IDs, occlusion handling |
| ROI | User-defined polygon — only vehicles inside ROI are analyzed |
| Congestion | 3 levels: 🟢 LOW / 🟡 MEDIUM / 🔴 HIGH (configurable thresholds) |
| Alerts | HIGH-congestion alerts with de-duplication + evidence snapshots |
| Outputs | Annotated MP4 · CSV report · Evidence JPGs · Console summary |

**How to use:**
1. Run all cells top-to-bottom (`Runtime → Run all`).
2. Upload your CCTV `.mp4` when prompted (or a sample video is downloaded automatically).
3. Optionally adjust the ROI polygon and congestion thresholds in the **Configuration** cell.
4. Collect results from the `Project/Outputs/` folder.

> 💡 For best speed: `Runtime → Change runtime type → GPU (T4)`.

## 5️⃣ Configuration — ROI Polygon, Thresholds & Visual Design

```python
# ==============================================================
# SECTION 5: SYSTEM CONFIGURATION
# ==============================================================

# ---------------- Detection classes (COCO IDs) ----------------
# Only these 5 vehicle classes are detected; everything else ignored.
VEHICLE_CLASSES = {
    1: "Bicycle",
    2: "Car",
    3: "Motorcycle",
    5: "Bus",
    7: "Truck",
}
CLASS_IDS = list(VEHICLE_CLASSES.keys())
CONF_THRESHOLD = 0.30          # minimum detection confidence
IOU_THRESHOLD  = 0.50          # NMS IoU threshold

# ---------------- Congestion thresholds (CONFIGURABLE) --------
# Based on number of vehicles currently inside the ROI.
CONGESTION_THRESHOLDS = {
    "LOW":    (0, 15),             # 0–15 vehicles  → LOW
    "MEDIUM": (16, 30),            # 16–30 vehicles → MEDIUM
    "HIGH":   (31, float("inf")),  # >30 vehicles   → HIGH
}

# ---------------- Professional color scheme (BGR) -------------
CONGESTION_COLORS = {
    "LOW":    (80, 200, 80),    # Green
    "MEDIUM": (0, 210, 255),    # Yellow
    "HIGH":   (60, 60, 230),    # Red
}
CLASS_COLORS = {                # unique bounding-box color per class
    "Car":        (255, 170, 0),    # blue
    "Bus":        (0, 140, 255),    # orange
    "Truck":      (200, 0, 200),    # magenta
    "Motorcycle": (0, 230, 230),    # yellow-cyan
    "Bicycle":    (180, 255, 100),  # light green
}

# ---------------- ROI polygon (user-defined) -------------------
# Defined as FRACTIONS of frame size (x, y) so it adapts to any
# resolution. Edit these points to match your camera view, e.g. a
# trapezoid covering the road area. Set ROI_FRACTION = None to
# analyze the FULL frame.
ROI_FRACTION = [
    (0.05, 0.98),   # bottom-left
    (0.95, 0.98),   # bottom-right
    (0.80, 0.25),   # top-right
    (0.20, 0.25),   # top-left
]

# ---------------- Performance options --------------------------
FRAME_SKIP   = 1        # 1 = process every frame; 2 = every 2nd frame, etc.
MAX_FRAMES   = None     # limit frames for quick tests (None = full video)
SNAPSHOT_COOLDOWN_SEC = 5.0   # min seconds between evidence snapshots
ALERT_COOLDOWN_SEC    = 10.0  # min gap between separate HIGH alert events

print("✅ Configuration loaded.")
print(f"   Vehicle classes : {list(VEHICLE_CLASSES.values())}")
print(f"   Thresholds      : LOW 0-15 | MEDIUM 16-30 | HIGH >30 (vehicles in ROI)")
print(f"   Frame skip      : {FRAME_SKIP} | Device: {DEVICE} | FP16: {USE_HALF}")
```

## 6️⃣ Core Functions — Model Loading, Detection & Tracking

```python
# ==============================================================
# SECTION 6A: MODEL LOADING & DETECTION / TRACKING FUNCTIONS
# ==============================================================

def load_model(model_path: Path = MODEL_PATH) -> YOLO:
    """Load YOLOv8s. Downloads weights if missing, stores them in Project/Models/."""
    try:
        if not model_path.exists():
            print("⬇️  yolov8s.pt not found locally — downloading from Ultralytics...")
            _ = YOLO("yolov8s.pt")             # auto-downloads to cwd
            src = Path("yolov8s.pt")
            if src.exists():
                shutil.move(str(src), str(model_path))
            else:
                raise FileNotFoundError("Model download did not produce yolov8s.pt")
        model = YOLO(str(model_path))
        model.to(DEVICE)
        # warm-up inference for stable FPS measurement
        model.predict(np.zeros((640, 640, 3), dtype=np.uint8),
                      verbose=False, device=DEVICE, half=USE_HALF)
        print(f"✅ Model loaded: {model_path} on {DEVICE.upper()}"
              + (" (FP16)" if USE_HALF else ""))
        return model
    except Exception as e:
        raise RuntimeError(f"❌ Failed to load YOLOv8 model: {e}")


def detect_objects(model: YOLO, frame: np.ndarray) -> sv.Detections:
    """Run YOLOv8 on one frame; keep only the 5 vehicle classes.
       Returns empty detections gracefully on failure."""
    try:
        results = model.predict(
            frame,
            classes=CLASS_IDS,          # detect ONLY vehicle classes
            conf=CONF_THRESHOLD,
            iou=IOU_THRESHOLD,
            device=DEVICE,
            half=USE_HALF,              # FP16 on GPU
            verbose=False,
        )[0]
        return sv.Detections.from_ultralytics(results)
    except Exception as e:
        print(f"⚠️ Detection error (frame skipped): {e}")
        return sv.Detections.empty()


def track_objects(tracker: sv.ByteTrack, detections: sv.Detections) -> sv.Detections:
    """Update ByteTrack with new detections.
       - assigns a persistent unique tracker_id to every vehicle
       - keeps IDs stable across frames (prevents duplicate counting)
       - lost_track_buffer handles temporary occlusions"""
    try:
        return tracker.update_with_detections(detections)
    except Exception as e:
        print(f"⚠️ Tracking error: {e}")
        return sv.Detections.empty()

print("✅ Model / detection / tracking functions defined.")
```

```python
# ==============================================================
# SECTION 6B: ROI, DENSITY & CONGESTION FUNCTIONS
# ==============================================================

def build_roi_polygon(width: int, height: int) -> np.ndarray:
    """Convert fractional ROI points into an absolute pixel polygon."""
    if ROI_FRACTION is None:  # full-frame analysis
        return np.array([[0, 0], [width - 1, 0],
                         [width - 1, height - 1], [0, height - 1]], dtype=np.int32)
    return np.array([(int(x * width), int(y * height))
                     for x, y in ROI_FRACTION], dtype=np.int32)


def filter_roi(detections: sv.Detections, roi_polygon: np.ndarray) -> sv.Detections:
    """Keep only vehicles whose bottom-center anchor lies INSIDE the ROI polygon.
       Vehicles outside the ROI are ignored completely."""
    if len(detections) == 0:
        return detections
    mask = []
    for xyxy in detections.xyxy:
        x1, y1, x2, y2 = xyxy
        anchor = (float((x1 + x2) / 2), float(y2))   # bottom-center point
        inside = cv2.pointPolygonTest(roi_polygon, anchor, False) >= 0
        mask.append(inside)
    return detections[np.array(mask, dtype=bool)]


def calculate_density(vehicle_count: int, roi_polygon: np.ndarray,
                      frame_area: int) -> float:
    """Vehicle density = vehicles per 10% of frame area covered by the ROI
       (a resolution-independent, normalized metric)."""
    roi_area = max(cv2.contourArea(roi_polygon), 1.0)
    return round(vehicle_count / (roi_area / frame_area * 10.0 + 1e-9), 2)


def estimate_congestion(vehicles_in_roi: int) -> str:
    """Map the current in-ROI vehicle count to LOW / MEDIUM / HIGH."""
    for level, (lo, hi) in CONGESTION_THRESHOLDS.items():
        if lo <= vehicles_in_roi <= hi:
            return level
    return "HIGH"

print("✅ ROI / density / congestion functions defined.")
```

## 7️⃣ Visualization — ROI, Bounding Boxes & Dashboard Overlay

```python
# ==============================================================
# SECTION 7: DRAWING FUNCTIONS
# ==============================================================
FONT = cv2.FONT_HERSHEY_SIMPLEX

def draw_roi(frame: np.ndarray, roi_polygon: np.ndarray, level: str) -> np.ndarray:
    """Draw the ROI polygon (semi-transparent fill tinted by congestion level)."""
    overlay = frame.copy()
    color = CONGESTION_COLORS[level]
    cv2.fillPoly(overlay, [roi_polygon], color)
    frame = cv2.addWeighted(overlay, 0.12, frame, 0.88, 0)
    cv2.polylines(frame, [roi_polygon], isClosed=True, color=color, thickness=3)
    label_pt = (int(roi_polygon[0][0]) + 5, int(roi_polygon[0][1]) - 8)
    cv2.putText(frame, "ROI", label_pt, FONT, 0.7, color, 2, cv2.LINE_AA)
    return frame


def draw_detections(frame: np.ndarray, detections: sv.Detections) -> np.ndarray:
    """Draw bounding box + class name + tracking ID + confidence per vehicle."""
    if len(detections) == 0:
        return frame
    for i in range(len(detections)):
        x1, y1, x2, y2 = detections.xyxy[i].astype(int)
        cls_id  = int(detections.class_id[i])
        conf    = float(detections.confidence[i])
        tid     = (int(detections.tracker_id[i])
                   if detections.tracker_id is not None else -1)
        name    = VEHICLE_CLASSES.get(cls_id, "Vehicle")
        color   = CLASS_COLORS.get(name, (255, 255, 255))

        cv2.rectangle(frame, (x1, y1), (x2, y2), color, 2)
        label = f"#{tid} {name} {conf:.2f}"
        (tw, th), _ = cv2.getTextSize(label, FONT, 0.5, 1)
        cv2.rectangle(frame, (x1, max(y1 - th - 8, 0)), (x1 + tw + 6, y1), color, -1)
        cv2.putText(frame, label, (x1 + 3, max(y1 - 5, th)),
                    FONT, 0.5, (0, 0, 0), 1, cv2.LINE_AA)
    return frame


def draw_dashboard(frame: np.ndarray, stats: dict) -> np.ndarray:
    """Draw the analytics dashboard panel + alert banner on every frame."""
    h, w = frame.shape[:2]
    level = stats["congestion"]
    lvl_color = CONGESTION_COLORS[level]

    # ---------- left panel background ----------
    panel_w, panel_h = 330, 345
    overlay = frame.copy()
    cv2.rectangle(overlay, (10, 10), (10 + panel_w, 10 + panel_h), (25, 25, 25), -1)
    frame = cv2.addWeighted(overlay, 0.72, frame, 0.28, 0)
    cv2.rectangle(frame, (10, 10), (10 + panel_w, 10 + panel_h), lvl_color, 2)

    # ---------- header ----------
    cv2.putText(frame, "TRAFFIC ANALYTICS", (24, 40), FONT, 0.72, (255, 255, 255), 2, cv2.LINE_AA)
    cv2.line(frame, (24, 50), (panel_w - 10, 50), (120, 120, 120), 1)

    rows = [
        (f"Cars        : {stats['Car']}",        CLASS_COLORS["Car"]),
        (f"Buses       : {stats['Bus']}",        CLASS_COLORS["Bus"]),
        (f"Trucks      : {stats['Truck']}",      CLASS_COLORS["Truck"]),
        (f"Motorcycles : {stats['Motorcycle']}", CLASS_COLORS["Motorcycle"]),
        (f"Bicycles    : {stats['Bicycle']}",    CLASS_COLORS["Bicycle"]),
        (f"In ROI now  : {stats['in_roi']}",     (255, 255, 255)),
        (f"Total (uniq): {stats['total_unique']}", (255, 255, 255)),
        (f"Density     : {stats['density']}",    (200, 200, 200)),
        (f"FPS         : {stats['fps']:.1f}",    (200, 200, 200)),
        (f"Frame       : {stats['frame_no']}",   (200, 200, 200)),
        (f"Time        : {stats['timestamp']}",  (200, 200, 200)),
    ]
    y = 78
    for text, color in rows:
        cv2.putText(frame, text, (24, y), FONT, 0.55, color, 1, cv2.LINE_AA)
        y += 24

    # ---------- congestion level badge ----------
    cv2.rectangle(frame, (24, y - 8), (panel_w - 14, y + 22), lvl_color, -1)
    txt_color = (255, 255, 255) if level == "HIGH" else (0, 0, 0)
    cv2.putText(frame, f"CONGESTION: {level}", (32, y + 14),
                FONT, 0.62, txt_color, 2, cv2.LINE_AA)

    # ---------- HIGH-congestion alert banner ----------
    if stats["alert_active"]:
        banner_h = 46
        overlay = frame.copy()
        cv2.rectangle(overlay, (0, h - banner_h), (w, h), (40, 40, 220), -1)
        frame = cv2.addWeighted(overlay, 0.85, frame, 0.15, 0)
        msg = "!! HIGH TRAFFIC CONGESTION !!"
        (tw, _), _ = cv2.getTextSize(msg, FONT, 0.95, 2)
        cv2.putText(frame, msg, ((w - tw) // 2, h - 14),
                    FONT, 0.95, (255, 255, 255), 2, cv2.LINE_AA)
    return frame

print("✅ Drawing functions defined.")
```

## 8️⃣ Alerts, Evidence Snapshots & CSV Reporting

```python
# ==============================================================
# SECTION 8: ALERTS, EVIDENCE & CSV REPORT FUNCTIONS
# ==============================================================

class AlertManager:
    """Generates HIGH-congestion alerts without duplicates.
       An 'alert event' starts when congestion becomes HIGH and ends
       when it drops back — one alert per continuous HIGH period."""
    def __init__(self):
        self.active = False              # currently inside a HIGH event?
        self.total_alerts = 0
        self.alert_log = []              # list of dicts (timestamp, frame)
        self.last_snapshot_t = -1e9
        self.last_alert_t = -1e9

    def update(self, level: str, frame_no: int, video_time: float, timestamp: str):
        """Returns (new_alert_triggered, snapshot_due)."""
        new_alert, snapshot_due = False, False
        if level == "HIGH":
            if not self.active and (video_time - self.last_alert_t) >= ALERT_COOLDOWN_SEC:
                # New HIGH-congestion event begins -> generate ONE alert
                self.active = True
                self.total_alerts += 1
                self.last_alert_t = video_time
                self.alert_log.append({"frame": frame_no, "timestamp": timestamp})
                new_alert = True
                print(f"⚠️  ALERT #{self.total_alerts}: HIGH TRAFFIC CONGESTION "
                      f"@ frame {frame_no} ({timestamp})")
            elif not self.active:
                self.active = True       # still within cooldown: no duplicate alert
            # snapshot throttling within an ongoing HIGH event
            if (video_time - self.last_snapshot_t) >= SNAPSHOT_COOLDOWN_SEC:
                self.last_snapshot_t = video_time
                snapshot_due = True
        else:
            self.active = False          # event ended -> next HIGH is a new alert
        return new_alert, snapshot_due


def generate_alert(alert_mgr: AlertManager, level: str, frame_no: int,
                   video_time: float, timestamp: str):
    """Wrapper: update the alert state machine for this frame."""
    return alert_mgr.update(level, frame_no, video_time, timestamp)


def save_snapshot(frame: np.ndarray, alert_idx: int, frame_no: int,
                  timestamp: str, level: str, vehicle_count: int) -> str:
    """Save an annotated evidence snapshot with metadata stamped on it."""
    try:
        existing = len(list(EVIDENCE_DIR.glob("*.jpg")))
        path = EVIDENCE_DIR / f"alert_{existing + 1:03d}.jpg"
        snap = frame.copy()
        meta = [f"EVIDENCE (alert #{alert_idx:03d})",
                f"Frame: {frame_no}",
                f"Time : {timestamp}",
                f"Level: {level}",
                f"Vehicles in ROI: {vehicle_count}"]
        y = max(snap.shape[0] - 140, 20)
        for line in meta:
            cv2.putText(snap, line, (max(snap.shape[1] - 380, 10), y),
                        FONT, 0.6, (255, 255, 255), 2, cv2.LINE_AA)
            y += 24
        cv2.imwrite(str(path), snap, [cv2.IMWRITE_JPEG_QUALITY, 92])
        return str(path)
    except Exception as e:
        print(f"⚠️ Snapshot save failed: {e}")
        return ""


def save_csv(rows: list, path: Path = CSV_REPORT_PATH):
    """Write the full per-frame traffic report to CSV."""
    header = ["Frame Number", "Timestamp", "Cars", "Buses", "Trucks",
              "Motorcycles", "Bicycles", "Total Vehicles",
              "Congestion Level", "Alert Generated", "Snapshot Path"]
    try:
        with open(path, "w", newline="") as f:
            writer = csv.writer(f)
            writer.writerow(header)
            writer.writerows(rows)
        print(f"✅ CSV report saved → {path}  ({len(rows)} rows)")
    except Exception as e:
        print(f"❌ Failed to write CSV report: {e}")

print("✅ Alert / evidence / CSV functions defined.")
```

```python
# ==============================================================
# SECTION 9: EFFICIENT OUTPUT VIDEO WRITING
# ==============================================================

def write_output_video(width: int, height: int, fps: float,
                       path: Path = OUTPUT_VIDEO_PATH) -> cv2.VideoWriter:
    """Create an MP4 writer matching the INPUT resolution & FPS."""
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")
    writer = cv2.VideoWriter(str(path), fourcc, fps, (width, height))
    if not writer.isOpened():
        raise IOError("❌ Could not open VideoWriter — check codec support.")
    return writer

print("✅ Video writer function defined.")
```

## 🔟 Main Processing Pipeline

```python
# ==============================================================
# SECTION 10: MAIN PROCESSING PIPELINE
# --------------------------------------------------------------
# read frame → detect → track → ROI filter → count → density →
# congestion → alerts → evidence → dashboard → write video → CSV
# ==============================================================

def process_video(model: YOLO, source) -> dict:
    """Run the full traffic-analysis pipeline over a video source."""

    cap = cv2.VideoCapture(source if isinstance(source, int) else str(source))
    if not cap.isOpened():
        raise IOError(f"❌ Cannot open video source: {source}")

    width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    src_fps = cap.get(cv2.CAP_PROP_FPS)
    src_fps = src_fps if src_fps and src_fps > 0 else 25.0
    total_src_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    roi_polygon = build_roi_polygon(width, height)
    frame_area = width * height

    # ByteTrack: lost_track_buffer keeps IDs alive through occlusions
    tracker = sv.ByteTrack(
        track_activation_threshold=0.25,
        lost_track_buffer=int(src_fps * 2),    # ~2 s occlusion tolerance
        minimum_matching_threshold=0.8,
        frame_rate=int(src_fps),
    )

    out_fps = src_fps / max(FRAME_SKIP, 1)
    writer = write_output_video(width, height, out_fps)

    alert_mgr = AlertManager()
    csv_rows = []
    unique_ids_by_class = defaultdict(set)   # class -> set of tracker IDs
    all_unique_ids = set()                   # prevents duplicate counting

    frame_no = 0
    processed = 0
    max_in_roi = 0
    level_history = []
    fps_smooth = 0.0
    t_start = time.time()
    corrupted = 0

    print(f"\n🚀 Processing started ({width}x{height} @ {src_fps:.1f} FPS, "
          f"device={DEVICE}, half={USE_HALF})\n" + "=" * 60)

    while True:
        # ---- 0. Read next frame (corrupted-frame tolerant) ----
        try:
            ret, frame = cap.read()
        except Exception:
            ret, frame = False, None
        if not ret or frame is None:
            break
        frame_no += 1

        if MAX_FRAMES and processed >= MAX_FRAMES:
            break
        if (frame_no - 1) % max(FRAME_SKIP, 1) != 0:   # frame-skip optimization
            continue
        if frame.size == 0 or frame.ndim != 3:          # corrupted frame guard
            corrupted += 1
            continue

        t0 = time.time()

        # ---- 1. Detect (vehicle classes only) ----
        detections = detect_objects(model, frame)
        # ---- 2. Track (persistent unique IDs) ----
        detections = track_objects(tracker, detections)
        # ---- 3. ROI filter (ignore vehicles outside ROI) ----
        detections = filter_roi(detections, roi_polygon)

        # ---- 4. Count vehicles (per class + total + unique) ----
        per_class_now = {name: 0 for name in VEHICLE_CLASSES.values()}
        if len(detections) > 0 and detections.tracker_id is not None:
            for cls_id, tid in zip(detections.class_id, detections.tracker_id):
                name = VEHICLE_CLASSES.get(int(cls_id), "Vehicle")
                per_class_now[name] += 1
                unique_ids_by_class[name].add(int(tid))   # unique cumulative
                all_unique_ids.add(int(tid))
        in_roi = len(detections)
        max_in_roi = max(max_in_roi, in_roi)

        # ---- 5. Density & congestion level ----
        density = calculate_density(in_roi, roi_polygon, frame_area)
        level = estimate_congestion(in_roi)
        level_history.append(level)

        # ---- 6. Timing / FPS analytics ----
        video_time = frame_no / src_fps
        timestamp = (datetime(2000, 1, 1) + timedelta(seconds=video_time)).strftime("%H:%M:%S")
        wall_ts = datetime.now().strftime("%H:%M:%S")
        proc_dt = time.time() - t0
        inst_fps = 1.0 / max(proc_dt, 1e-6)
        fps_smooth = inst_fps if fps_smooth == 0 else 0.9 * fps_smooth + 0.1 * inst_fps

        # ---- 7. Alert logic (de-duplicated) ----
        new_alert, snapshot_due = generate_alert(alert_mgr, level, frame_no,
                                                 video_time, timestamp)

        # ---- 8. Annotate frame ----
        annotated = draw_roi(frame, roi_polygon, level)
        annotated = draw_detections(annotated, detections)
        stats = {
            **per_class_now,
            "in_roi": in_roi,
            "total_unique": len(all_unique_ids),
            "density": density,
            "congestion": level,
            "fps": fps_smooth,
            "frame_no": frame_no,
            "timestamp": wall_ts,
            "alert_active": alert_mgr.active,
        }
        annotated = draw_dashboard(annotated, stats)

        # ---- 9. Evidence snapshot on HIGH congestion ----
        snapshot_path = ""
        if snapshot_due:
            snapshot_path = save_snapshot(annotated, alert_mgr.total_alerts,
                                          frame_no, timestamp, level, in_roi)

        # ---- 10. Write annotated output frame ----
        writer.write(annotated)

        # ---- 11. Record CSV row ----
        csv_rows.append([
            frame_no, timestamp,
            per_class_now["Car"], per_class_now["Bus"], per_class_now["Truck"],
            per_class_now["Motorcycle"], per_class_now["Bicycle"],
            in_roi, level, "Yes" if new_alert else "No", snapshot_path,
        ])

        processed += 1
        if processed % 50 == 0:
            pct = f"{frame_no / total_src_frames * 100:.0f}%" if total_src_frames > 0 else "live"
            print(f"  frame {frame_no:>6} ({pct:>4}) | in-ROI: {in_roi:>3} | "
                  f"level: {level:<6} | fps: {fps_smooth:5.1f}")

    cap.release()
    writer.release()
    total_time = time.time() - t_start

    save_csv(csv_rows)

    return {
        "processed": processed,
        "total_time": total_time,
        "avg_fps": processed / max(total_time, 1e-6),
        "max_in_roi": max_in_roi,
        "total_alerts": alert_mgr.total_alerts,
        "alert_log": alert_mgr.alert_log,
        "level_history": level_history,
        "unique_by_class": {k: len(v) for k, v in unique_ids_by_class.items()},
        "total_unique": len(all_unique_ids),
        "corrupted": corrupted,
    }

print("✅ Main pipeline defined.")
```

## 1️⃣1️⃣ Run the System

```python
# ==============================================================
# SECTION 11: MAIN ENTRY POINT + FINAL SUMMARY
# ==============================================================

def main():
    """Load the model, run the full pipeline, print the final summary."""
    model = load_model()
    results = process_video(model, INPUT_VIDEO_PATH)

    # ---------------- FINAL SUMMARY ----------------
    levels = results["level_history"]
    if levels:
        score = {"LOW": 0, "MEDIUM": 1, "HIGH": 2}
        avg_score = sum(score[l] for l in levels) / len(levels)
        avg_level = ("LOW" if avg_score < 0.5 else
                     "MEDIUM" if avg_score < 1.5 else "HIGH")
        dist = {l: f"{levels.count(l)/len(levels)*100:.1f}%"
                for l in ("LOW", "MEDIUM", "HIGH")}
    else:
        avg_level, dist = "N/A", {}

    print("\n" + "=" * 60)
    print("                 📊 FINAL SUMMARY")
    print("=" * 60)
    print(f"  Total Frames Processed   : {results['processed']}")
    print(f"  Processing Time          : {results['total_time']:.1f} s")
    print(f"  Average FPS              : {results['avg_fps']:.2f}")
    print(f"  Maximum Vehicle Count    : {results['max_in_roi']} (in ROI at once)")
    print(f"  Total Unique Vehicles    : {results['total_unique']}")
    for cls, n in results["unique_by_class"].items():
        print(f"     - {cls:<11}: {n}")
    print(f"  Total Alerts Generated   : {results['total_alerts']}")
    for a in results["alert_log"]:
        print(f"     - frame {a['frame']} @ {a['timestamp']}")
    print(f"  Average Congestion Level : {avg_level}  {dist}")
    print(f"  Corrupted Frames Skipped : {results['corrupted']}")
    print("-" * 60)
    print(f"  ✅ Output video : {OUTPUT_VIDEO_PATH}")
    print(f"  ✅ CSV report   : {CSV_REPORT_PATH}")
    n_evidence = len(list(EVIDENCE_DIR.glob('*.jpg')))
    print(f"  ✅ Evidence     : {n_evidence} snapshot(s) in {EVIDENCE_DIR}/")
    print("=" * 60)
    return results

# ---------------- RUN THE SYSTEM ----------------
results = main()
```
