```python
import cv2
import os
import time

# ==========================
# Create output folder
# ==========================
save_folder = "captured_parking_area"
os.makedirs(save_folder, exist_ok=True)

# ==========================
# Initialize Camera
# ==========================
print("[INFO] Initializing camera...")

camera = cv2.VideoCapture(0, cv2.CAP_DSHOW)

camera.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
camera.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)

if not camera.isOpened():
    print("[ERROR] Failed to initialize camera.")
    exit()

print("[INFO] Camera started successfully.")

time.sleep(2)

print("\n")
print("===================================")
print("Press SPACE to capture image")
print("Press Q to quit")
print("===================================")

count = 1

while True:

    ret, frame = camera.read()

    if not ret:
        print("[ERROR] Could not read frame.")
        break

    display = frame.copy()

    cv2.putText(
        display,
        "SPACE = Capture | Q = Quit",
        (20,35),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0,255,0),
        2
    )

    cv2.imshow("Parking Area Capture", display)

    key = cv2.waitKey(1) & 0xFF

    # Quit
    if key == ord('q'):
        print("[INFO] Camera closed.")
        break

    # Capture Image
    elif key == 32:

        filename = os.path.join(
            save_folder,
            f"parking_area_{count}.jpg"
        )

        cv2.imwrite(filename, frame)

        print(f"[INFO] Image saved: {filename}")

        count += 1

camera.release()
cv2.destroyAllWindows()
video_filename = input("Video name : ")

img_name = f"{video_filename}.jpg"

txt_name = f"{video_filename}.txt"

print("[INFO] Finished.")
```
