```python
import os
import cv2
import numpy as np
from ultralytics import YOLO
import matplotlib.pyplot as plt
```

```python
from ultralytics import YOLO

# Person + Vehicle Detection
vehicle_model = YOLO("models/yolov8s.pt")

# Parking Slot Detection
parking_model = YOLO("models/parking_best.pt")
```

```python
import os
output_folder = "output_videos"

os.makedirs(output_folder, exist_ok=True)


print("Output Folder Ready")
```

```python
from ultralytics import YOLO

parking_model = YOLO("models/parking_best.pt")
print(parking_model.names)
```

```python
import os

input_folder = "input_videos"

videos = [f for f in os.listdir(input_folder)
          if f.endswith((".mp4", ".avi", ".mov", ".mkv"))]

print("Available Videos:")

for i, video in enumerate(videos):
    print(f"{i+1}. {video}")

choice = int(input("\nEnter Video Number: "))

video_path = os.path.join(input_folder, videos[choice-1])

print("\nSelected:", video_path)
```

```python
video_name = os.path.basename(video_path)

output_path = os.path.join(
    output_folder,
    video_name.replace(".mp4", "_output.mp4")
)

cap = cv2.VideoCapture(video_path)
print("Video Opened:", cap.isOpened())

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

new_width = 1280
new_height = 720

out = cv2.VideoWriter(
    output_path,
    fourcc,
    fps,
    (new_width, new_height)
)

print("Output Path:", output_path)
print("FPS:", fps)
print("Width:", width)
print("Height:", height)
print("VideoWriter Open:", out.isOpened())
cv2.namedWindow("Parking Detection", cv2.WINDOW_NORMAL)
cv2.resizeWindow("Parking Detection", 1280, 720)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    cv2.imshow("Parking Detection", annotated_frame)

    key = cv2.waitKey(1)

    if key == ord('q'):
        break

    frame = cv2.resize(frame, (1280, 720))

    # YOLO Detection
    vehicle_results = vehicle_model(
        frame,
        imgsz=640,
        conf=0.55,
        classes=[0,2,5,7],
        verbose=False
    )

    parking_results = parking_model(
        frame,
        imgsz=640,
        conf=0.35,
        verbose=False
    )
    annotated_frame = frame.copy()
    # ---------------- Draw Boxes ----------------
person_boxes = []
vehicle_boxes = []
vehicle_centers = []

for box in vehicle_results[0].boxes:

    cls = int(box.cls[0])
    conf = float(box.conf[0])

    x1, y1, x2, y2 = map(int, box.xyxy[0])

    cx = (x1 + x2) // 2
    cy = (y1 + y2) // 2

    if cls == 0:

        label = f"Person {conf:.2f}"
        color = (0, 255, 0)

        person_boxes.append((x1, y1, x2, y2))

    else:

        label = f"Vehicle {conf:.2f}"
        color = (255, 0, 0)

        vehicle_boxes.append((x1, y1, x2, y2))
        vehicle_centers.append((cx, cy))

    cv2.rectangle(
        annotated_frame,
        (x1, y1),
        (x2, y2),
        color,
        2
    )

    cv2.putText(
        annotated_frame,
        label,
        (x1, y1 - 10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        color,
        2
    )
    # ---------------- Counts ----------------
    person_count = len(person_boxes)
    vehicle_count = len(vehicle_boxes)

    alert = "SAFE"

    if person_count > 0:
        alert = "WARNING : Person Detected"

    if person_count > 5:
        alert = "HIGH CROWD"

    # ---------------- Person-Vehicle Distance ----------------
    for p in person_boxes:

        px = (p[0] + p[2]) // 2
        py = (p[1] + p[3]) // 2

        for v in vehicle_boxes:

            vx = (v[0] + v[2]) // 2
            vy = (v[1] + v[3]) // 2

            distance = np.sqrt((px - vx) ** 2 + (py - vy) ** 2)

            if distance < 120:

                alert = "ALERT : Person Near Vehicle"

                cv2.line(
                    annotated_frame,
                    (px, py),
                    (vx, vy),
                    (0, 0, 255),
                    2
                )
vacant = 0
occupied = 0

for box in parking_results[0].boxes:

    cls = int(box.cls[0])
    conf = float(box.conf[0])

    x1, y1, x2, y2 = map(int, box.xyxy[0])

    if cls == 0:

        color = (0,255,0)
        label = f"Vacant {conf:.2f}"
        vacant += 1

    else:

        color = (0,0,255)
        label = f"Occupied {conf:.2f}"
        occupied += 1

    cv2.rectangle(
        annotated_frame,
        (x1,y1),
        (x2,y2),
        color,
        2
    )

    cv2.putText(
        annotated_frame,
        label,
        (x1,y1-10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        color,
        2
    )

    # ---------------- Dashboard ----------------
    cv2.putText(
        annotated_frame,
        f"Persons : {person_count}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0, 255, 0),
        2
    )

    cv2.putText(
        annotated_frame,
        f"Vehicles : {vehicle_count}",
        (20, 80),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (255, 0, 0),
        2
    )

    cv2.putText(
        annotated_frame,
        alert,
        (20, 120),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0, 0, 255),
        2
    )
cv2.putText(
    annotated_frame,
    f"Occupied : {occupied}",
    (20,160),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.8,
    (0,0,255),
    2
)

cv2.putText(
    annotated_frame,
    f"Vacant : {vacant}",
    (20,200),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.8,
    (0,255,0),
    2
)


      # ---------------- Save & Display ---------------

out.write(annotated_frame)

cv2.imshow("Parking Detection", annotated_frame)

key = cv2.waitKey(1) & 0xFF

if key == ord('q'):
    break

try:
    if cv2.getWindowProperty("Parking Detection", cv2.WND_PROP_VISIBLE) < 1:
        break
except:
    break

# ---------------- Release ----------------
cap.release()
out.release()
cv2.destroyAllWindows()

print("=" * 50)
print("Detection Completed")
print("Saved To :", output_path)
print("=" * 50)
```
